# 🧮 Sets in Data Structure (Python)

## 📌 Topic Overview
A set is an unordered collection of unique elements in Python. It automatically removes duplicates and supports mathematical set operations like union, intersection, and difference.

### Real-world analogy:
Think of a set like a guest list for a party — each person can appear only once, no duplicates allowed.


### 🧠 Key Concepts

#### Set
- Unordered collection of unique elements

#### Mutable
- Elements can be added or removed

####  No Indexing
- Cannot access elements via index like lists

####  Hashable Elements
- Only immutable (hashable) elements allowed (e.g., int, tuple)

####  Set Operations
- Union (|)
- Intersection (&)
- Difference (-)
- Symmetric Difference (^)

####  Frozen Set
- Immutable version of a set

### 🧾 Syntax / Structure

```python
# Creating a set
my_set = {1, 2, 3, 4}

# Empty set (important)
empty_set = set()

# Set with mixed types
mixed_set = {1, "hello", 3.14}

# From list (removes duplicates)
set_from_list = set([1, 2, 2, 3])
```

#### Explanation:
- {} → Used for creating sets (but {} alone creates a dictionary)
- set() → Used to create an empty set
- Automatically removes duplicates

### 💡 Beginner Examples

#### Example 1: Removing duplicates

```python
numbers = [1, 2, 2, 3, 4, 4]
unique_numbers = set(numbers)

print(unique_numbers)
```

#### Example 2: Adding elements

```python
my_set = {1, 2, 3}
my_set.add(4)

print(my_set)
```

#### Example 3: Removing elements

```python
my_set = {1, 2, 3}
my_set.remove(2)

print(my_set)
```

### 🚀 Intermediate Examples

#### Example 1: Union & Intersection

```python
A = {1, 2, 3}
B = {3, 4, 5}

print("Union:", A | B)
print("Intersection:", A & B)
```

#### Example 2: Finding common users

```python
users_app1 = {"Ajay", "Rahul", "Sneha"}
users_app2 = {"Sneha", "Amit", "Ajay"}

common_users = users_app1 & users_app2

print(common_users)
```

### 🔥 Advanced Concepts

#### 1. Frozen Set

```python
fs = frozenset([1, 2, 3])
# fs.add(4) ❌ Error
```

- Immutable version of set
- Can be used as dictionary keys

#### 2. Performance Considerations
- Average time complexity:
    - Add: O(1)
    - Remove: O(1)
    - Lookup: O(1)
- Uses hash table internally

#### 3. Internal Working
- Based on hashing
- Each element stored using a hash value
- Ensures fast lookup and uniqueness

#### 4. Edge Cases

```python
# Duplicate values ignored
s = {1, 1, 1}
print(s)  # {1}

# Unhashable type
# s = {[1, 2]} ❌ Error
```

## ⚠️ Common Mistakes

#### ❌ Mistake 1: Using {} for empty set

```python
s = {}
print(type(s))  # dict ❌
```

#### ✅ Correct

```python
s = set()
```

#### ❌ Mistake 2: Using list inside set

```python
s = {[1, 2]}  # ❌ Error
```

#### ✅ Correct

```python
s = {(1, 2)}  # tuple allowed
```

#### ❌ Mistake 3: Using remove() without check

```python
s = {1, 2}
s.remove(3)  # ❌ Error
```

#### ✅ Correct

```python
s.discard(3)  # No error
```

## 🎯 Interview Questions

1. What is the difference between set and list?
- Set: Unordered, unique elements
- List: Ordered, allows duplicates

2. Why are sets faster than lists?
- Due to hash table implementation
- O(1) lookup vs O(n) in lists

3. Difference between remove() and discard()?
- remove() → Raises error if element not found
- discard() → No error

4. Can sets contain mutable elements?
- ❌ No (only hashable/immutable)

5. What is frozenset?
- Immutable version of set

## 🧩 Mini Project

#### 🔹 Unique Visitor Tracker

```python
visitors = set()

def add_visitor(name):
    visitors.add(name)

def total_visitors():
    return len(visitors)

add_visitor("Ajay")
add_visitor("Rahul")
add_visitor("Ajay")

print("Unique Visitors:", total_visitors())
```
