# 📖 Topic Description

## A dictionary in Python is a collection of key-value pairs where each key is unique and maps to a value. It is unordered, mutable, and indexed by keys.

# 👉Real-world analogy: 
## Think of a dictionary like a phonebook — you search by name (key) to get the phone number (value).

# 🧠 Key Concepts
## Key
#### Unique identifier used to access a value
#### Must be immutable (e.g., string, number, tuple)

## value
#### Data associated with a key
#### Can be of any data type

## Key-Value Pair
#### Combination of key and its value

## Mutable
#### Dictionary can be modified after creation

## Unordered (before Python 3.7 conceptually)
#### No fixed order of elements (but now insertion order is preserved)

## Hashing
#### Keys are hashed internally for fast lookup

## 🧾 Syntax / Structure

```python
# Basic dictionary syntax 
my_dict = {
    "key1": "value1",
    "key2": "value2"
}
```

### Explanation:
#### {} → Defines a dictionary
#### "key" → Unique identifier
#### : → Separates key and value
#### , → Separates pairs

## Beginner Examples

### Example 1: Creating a Dictionary

```python
# Example 1: Creating a Dictionary
student = {
    "name": "Ajay",
    "age": 25,
    "course": "Python"
} 

print(student) # Output: {'name': 'Ajay', 'age': 25, 'course': 'Python'}
```

### Example 2: Accessing Values

```python
print(student["name"])  # Access using key
```

### Example 3: Adding & Updating Values

```python
student["age"] = 26        # Update
student["city"] = "Delhi"  # Add new key

print(student) # Output: {'name': 'Ajay', 'age': 26, 'course': 'Python', 'city': 'Delhi'}
```

## 🚀 Intermediate Examples

### Example 1: Loop Through Dictionary

```python
for key, value in student.items():
    print(key, ":", value)

# Output:
# name : Ajay    
# age : 26
# course : Python
# city : Delhi
```

### Example 2: Count Frequency of Elements

```python
numbers = [1, 2, 2, 3, 3, 3]
freq = {}

for num in numbers:
    freq[num] = freq.get(num, 0) + 1

print(freq)
```

### Example 3: Nested Dictionary

```python
employees = {
    "emp1": {"name": "A", "salary": 50000},
    "emp2": {"name": "B", "salary": 60000}
}

print(employees["emp1"]["name"]) # Output: A
```

## 🔥 Advanced Concepts

### 1. Dictionary Comprehension

```python
squares = {x: x*x for x in range(5)} # Dictionary comprehension to create a dictionary of squares
print(squares) # Output: {0: 0, 1: 1, 2: 4, 3: 9, 4: 16}
```

### 2. Handling Missing Keys

```python
d = {"a": 1} # Create a dictionary with one key-value pair

print(d.get("b", "Not Found"))  # Safe access with default value, Output: Not Found
```

### 3. Performance Considerations

#### Lookup, insert, delete → O(1) average
#### Uses hash table internally
#### Collisions handled via hashing techniques

### 4. Immutable Keys Only

```python
# ❌ Wrong
# d = {[1,2]: "value"}  # List is mutable

# ✅ Correct
d = {(1,2): "value"}
```

## ⚠️ Common Mistakes

### Mistake 1: Using Non-Existent Key

```python
# ❌ Wrong
print(student["salary"])  # KeyError

# ✅ Correct
print(student.get("salary", "Not Available"))
```

### Mistake 2: Duplicate Keys

```python
# ❌ Wrong
d = {"a": 1, "a": 2}

print(d)  # Only last value kept
```

### Mistake 3: Using Mutable Key

```python
# ❌ Wrong
d = {[1,2]: "value"}  # Error
```

## 🧩 Mini Project

### 📊 Student Management System

```python
students = {}

while True:
    name = input("Enter student name: ")
    marks = int(input("Enter marks: "))
    
    students[name] = marks
    
    choice = input("Add more? (y/n): ")
    if choice.lower() != 'y':
        break

print("\nStudent Records:")
for name, marks in students.items():
    print(name, ":", marks)
```
