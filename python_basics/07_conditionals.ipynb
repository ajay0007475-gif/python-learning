# 📘 Conditional Statements in Python

## 📌 Topic Overview

### Conditional statements allow a program to make decisions based on conditions. They control the flow of execution depending on whether a condition is True or False.

#### Real-world analogy:
#### Like a traffic signal —
- 👉 If light is green → go
- 👉 If red → stop

### 🧠 Key Concepts

#### Condition
- Expression that evaluates to True or False

#### if Statement
- Executes code if condition is True

#### elif (Else If)
- Checks another condition if previous is False

#### else Statement
- Executes when all conditions are False

#### Comparison Operators
- ==, !=, >, <, >=, <=

#### Logical Operators
- and, or, not

#### Nested Conditions
- Condition inside another condition

### 🧾 Syntax / Structure

```python
# Basic if
if condition:
    # code block

# if-else
if condition:
    # code
else:
    # code

# if-elif-else
if condition1:
    # code
elif condition2:
    # code
else:
    # code
```

#### Explanation:
- if → first condition check
- elif → additional conditions
- else → fallback block
- Indentation is mandatory in Python

## 💡 Beginner Examples

### Example 1: Simple if

```python
age = 18

if age >= 18:
    print("You are eligible to vote")
```

### Example 2: if-else

```python
number = 7

if number % 2 == 0:
    print("Even number")
else:
    print("Odd number")
```

### Example 3: if-elif-else

```python
marks = 75

if marks >= 90:
    print("Grade A")
elif marks >= 60:
    print("Grade B")
else:
    print("Grade C")
```

## 🚀 Intermediate Examples

### Example 1: Multiple Conditions

```python
age = 25
citizen = True

if age >= 18 and citizen:
    print("Eligible to vote")
else:
    print("Not eligible")
```

### Example 2: Login Validation

```python
username = "admin"
password = "1234"

if username == "admin" and password == "1234":
    print("Login Successful")
else:
    print("Invalid Credentials")
```

## 🔥 Advanced Concepts

### 1. Nested Conditions

```python
age = 20
has_id = True

if age >= 18:
    if has_id:
        print("Entry allowed")
    else:
        print("ID required")
else:
    print("Underage")
```

### 2. Ternary Operator (Short if-else)

```python
number = 10

result = "Even" if number % 2 == 0 else "Odd"
print(result)
```

### 3. Truthy & Falsy Values

```python
if 0:
    print("Will not run")

if "Hello":
    print("Runs because string is truthy")
```

### 4. Edge Cases

```python
x = None

if x:
    print("Has value")
else:
    print("No value")
```

## ⚠️ Common Mistakes

### ❌ Missing Indentation

```python
if True:
print("Hello")  # Error
```

### ✅ Correct

```python
if True:
    print("Hello")
```

### ❌ Using = instead of ==

```python
if x = 5:  # Error
```

### ✅ Correct

```python
if x == 5:
```

### ❌ Wrong condition order

```python
marks = 85

if marks >= 60:
    print("Pass")
elif marks >= 90:
    print("Topper")  # Never runs
```

## 🎯 Interview Questions

### 1. Difference between if and elif?
- if → first condition
- elif → checked only if previous fails
### 2. What is a ternary operator?
- Short form of if-else
- Used in one line
### 3. What are truthy and falsy values?
- Truthy → non-zero, non-empty
- Falsy → 0, None, "", False
### 4. Can we use multiple elif?
- ✅ Yes
### 5. Does Python support switch-case?
- ❌ Not traditionally (use if-elif or match-case in Python 3.10+)
